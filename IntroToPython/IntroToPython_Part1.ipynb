Now is a good time to go over the native mathematical operations available to you in Python (we've seen addition and multiplication so far).

```python
addition = 5 + 5
print("addition: ",addition)

subtraction = 5 - 7
print("subtraction: ",subtraction)

multiplication = 5 * 5
print("multiplication: ",multiplication)

division = 5 / 6 # be careful of Python version!!
print("division: ",division)

exponentiation = 5**2
print("exponentiation: ",exponentiation)

modulus = 10 % 3
print("modulus: ",modulus)
```

Take a close look at some of the particular symbols (several of which are known as **operators**) used above that have specific meanings in Python:
1. **Multiplication** in Python uses the * symbol, not the x---the letter acts like any other letter, whereas * is an *operator*
2. **Exponents** in Python use **, not the ^ symbol. This is easy to forget! The ^ symbol is an operator as well, but it has a special use called XOR ("exclusive or") in Python, which is useful for certain logical calculations. We will not use it much here.
3. The **modulus** operator % returns the *remainder* of a division operation (i.e., 3 goes into 10 three times with 1 left over, so 10 % 3 = 1)
4. The # symbol in Python is not an operator: it denotes a **comment**. Placing a # tells the Python interpreter to *ignore everything after that symbol on the same line*. This is very useful for leaving descriptions or instructions in your code to help a future reader (a collaborator, or a future version of yourself) remember what the code is doing or how to use it.

Speaking of comments, what did I mean by "Be careful of Python Version" in the comment above?

Your output from the mathematical operations above probably includes a line that says `division:  0.8333333`---if so, you are using Python 3. However, if you are using Python 2, however, you would get an output of `division:  0`, that is 5/6 = 0, which is probably not the result you expect!

The reason for this is related to the idea of a **data type**: every object in Python (e.g., anything that can be stored as a variable) has a **type**, and an object's **type** determines how certain operations affect it. Numbers with no decimal points are treated as **integers**, whereas numbers that include digits past the decimal point are called **floats** (floating point numbers). We can see this by checking the type of a few values with the `type()` function:

```python
print(type(5))
print(type(5.5))
print(type(5.))
```

We'll discuss why you should care about data types shortly, but an important thing to note is that in Python version 2.xx, mathematical operations between integer-type values *will always return an integer-type result*. Since 0.83333 is not an integer, Python 2 cuts the value off at the decimal point, leaving only the leading zero for the answer. If we want to mimic in Python 3 the integer behavior of Python 2, we can use the // operator (known as "floor division" or "integer division") to force Python to treat our result as an integer, or we can change the result into an integer after the calculation is completed:

```python
# calculate 5/6 in the default manner and display the result and type
x=5/6
print(x,type(x))

# calculate 5/6 using "floor division" and display the result and type
y=5//6
print(y,type(y))

# calculate 5/6, then make it an integer and display the result and type
z=int(5/6)
print(z,type(z))
```

Note that in the example above, `int()` is a **function** that tells Python to convert the result from a float into an integer, which is zero in this case. Functions are objects that take in one or more **arguments** (whatever you put in the parentheses) and return a result or perform some other operation. There are similar functions such as `float()`, which converts its argument to a float---converting an object's type in this way is called **typecasting**. Another useful function we've been using is the `print()` function, which prints the value of its argument to the screen.

## Order of Operations

One more thing to remember about mathematical operations is that Python follows the normal mathematical rules for the [order of mathematical operations](https://en.wikipedia.org/wiki/Order_of_operations) (often called [PEMDAS](https://en.wikipedia.org/wiki/Order_of_operations#Mnemonics) in the USA). Expressions within **P**arentheses are evaluated first, then **E**xponents, then **M**ultiplication and **D**ivision, and then **A**ddition and **S**ubtraction. When two operations have the same priority (e.g., multiplication and division, or repeated subtraction) then the order proceeds from left to right.

## Data Types in Python

As mentioned above, every object in Python has a type, and *an object's type determines how certain operations affect it*. So far, we have been working entirely with numbers (integers and floats), but there are other types of objects in Python as well:

```python
u = 5
v = 6.0
w = 2.31e4 # exponential notation: equivalent to 2.31 * 10**4
x = 1.5+4.2j # complex notation: 4.2j is 4.2 times the square root of -1 (note lack of * between 4.2 and j!!)
y = '7.0'
z = [8]
print(type(u))
print(type(v))
print(type(w))
print(type(x))
print(type(y))
print(type(z))
```

While most data we discuss in this class will be numbers like these, Python's power comes in when you start looking at  other data types, most of which are designed to contain numbers in various organized ways. Here are the basic data types in Python:

 * Integers
 * Floats
 * Complex numbers
 * Booleans (True or False)
 * Lists (ordered collections of items)
 * Dictionaries (unordered collections accessed via "keys")
 * Strings (sets of characters contained in quotes "like this")
 * Tuples (like lists, but *immutable*, i.e., unchangeable)

We will discuss all of these types over time, and we will also use other types of objects using packages that add additional power to Python. For Part 1 of this introduction, we will only discuss two of the above: **booleans** and **strings**.

```python
a = True
b = False
```

These come in handy when we are employing **conditional statements** (coming up in Part 2) in which we say "Hey code, if some condition is True, do "X," else if some other condition is True, do "Y". Often we are using booleans without recognizing it.

It is easy to confuse the `==` operator (which _compares_ values, rather than assigning them) and the `=` operator!

### Strings

Strings let you store things like words (or, later on, filenames and paths) in your code that otherwise don't have meaning as far as Python is concerned. Strings are the most forgiving data type of all, you can stick literally anything inside as long as you enclose the string with double quotes (") or single quotes (').

The downside of strings is that you cannot do math on them. Python does not interpret characters in strings as numbers. Instead, using normal math operators on strings will produce different effects. In addition, math operations *within* strings will not be evaluated at all, which is good if you want to print an expression to the screen that includes an operator.

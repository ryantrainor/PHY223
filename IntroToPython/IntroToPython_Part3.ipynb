#### What's a function?
It's useful to take a second to make sure we're on the same page about functions. Recall that a **function** is something that takes one or more inputs (or **arguments**), and then performs some operation on them (often returning a numerical value as output). When, in math class, you write $y = \sin(x)$, "sin" is the function you are using, and "x" is the argument of the function, and you are defining "y" based on this function of x.

If I want to calculate the sine of a number, I can't do that in native Python. But luckily, many clever people have crafted libraries like NumPy that  extend the basic Python capabilities. NumPy includes the `np.sin()` function, which returns the sine of any number (in radians) that it receives as an argument. Better yet, if you give `np.sin()` an argument that is an entire numpy array of values, the function will return an entire numpy array (equal in length to the input array) of output values: *it calculates the sine element-wise*.

Let's see this in action. First we'll need to **import** numpy, and then we can try out the sine function:

```python
# Run this cell and see what happens

import numpy as np

print(np.sin(0))
print(np.sin(1))
print(np.sin(np.pi/2))
```

A few notes:
* Notice that numpy includes some other helpful math tools like the value of pi (`np.pi`)

* We could have imported numpy with `import numpy` instead of `import numpy as np`. This would be faster in the short run, but then we would have to type `numpy.pi` every time we wanted to use pi, `numpy.sin()` everytime we wanted to use the sine function, etc. By assigning the nickname "np", we can use `np.pi` and `np.sin()` to save ourselves some typing. You can choose any nickname you want in your own code, but numpy is usually imported as `np`.

Next we will try using `np.sin()` on a whole array of numbers at once, but first we need to generate a numpy array of such values. Run the cell below to see what is produced:

```python
# Run this cell to generate an array of values so that we can take the sine of each one
x = np.linspace(0,10,101)
print(x)
```

Ok, what did we do here??

Firstly, note that we just introduced a new function for generating arrays of sequential numbers: `np.linspace()`. This function is somewhat similar to the `range()` function we used in Part 2, which also produced a series of sequential numbers. Recall that `range()` had you pick a starting value, a stopping value, and an (integer) step size; `np.linspace()` lets you pick a starting value, a stopping value, and a *total number of points* that are then spaced evenly between the endpoints. You can read `np.linspace(0,10,101)` as "give me an array from 0 to 10 with 101 evenly spaced points."

Why did we choose 101 evenly spaced points? That's a funny number, but notice that `np.linspace()` creates an array that is *inclusive* of both the initial and final number (i.e., it includes both 0 and 10), so we need 101 elements to have a separation of 0.1 between values. This is different from the behavior of `range()`, which ends just before the stopping value.

For what it's worth, there are a few other functions similar to `np.linspace()` that are helpful for generating arrays of values:
1. `np.arange()` is even more similar to the native Python `range` function. The syntax is `np.arange(start,stop,step)`, which is the same as `range` except that `np.arange` can use non-integer values for all three arguments, and it returns a numpy array instead of a python iterator object. Thus, `np.linspace(0,10,101)` is equivalent to `np.arange(0,10.1,0.1)` (note that `np.arange` acts like `range` in that it does not include the "stop" value). It can be useful to use `np.arange()` when we want to specify the spacing between our points, whereas it's useful to use `np.linspace` when we are more interested in specifying the total number.

2. `np.logspace()` creates values that are evenly spaced in logarithmic space (i.e., each array element increases by the same *multiplicative* factor over the previous element instead of increasing by the same *additive* factor). The arguments are powers of 10: for example, `np.logspace(-1,3,5)` will return an array with the elements 0.1, 1, 10, 100, 1000 (five values ranging from 10**-1 to 10**3). This is equivalent to `10**np.linspace(-1,3,5)`.

3. `np.zeros(dims)` produces an array of zeros with dimension `dims`. If `dims` is a single whole number, then the array is a 1D array with that length: `np.zeros(3)` is the array 0.0, 0.0, 0.0. If `dims` is a tuple of whole numbers, then the array has a number of dimensions equal to `len(dims)` with the values of the tuple as the length of each dimension: `np.zeros((3,6,7))` produces a 3D array (3 x 6 x 7), with all 126 elements being equal to 0.0.

4. `np.ones(dims)` is identical to `np.zeros(dims)` except that every element is equal to 1.0 instead of 0.0.

In any case, now we want to calculate the sine of each of the values in our `x` array and store them as a new array, `y`:

```python
# Take the sine of the values we generated previously
y = np.sin(x)
print(y)
```

The values also seem to gradually oscillate between -1 and 1, which is the expected behavior of a sine curve, so that's promising as well. We'll get to those things later, however, and this graph is already enough to clearly see the behavior of the function and the relationships between our `x` and `y` variables, which is why we usually make graphs.

There is one somewhat subtle issue that makes this plot different from other graphs you may have made of a sine curve in the past, however. Mathematically, we usually think of graphs like this as **continuous**: $\sin(x)$ is indeed a continuous function for all real values of $x$. However, our plot is based on our `x` and `y` arrays, which are collections of **discrete points**. We can see this more clearly if we mark the individual points in our plotted arrays with dots in our plot:


```python
# Plot the graph again with dots at each point

fig = plt.figure()
ax = fig.add_subplot(111)
ax.plot(x,y,marker='.')
```

It's hard to tell here, but the dots in the above graph are the only points where the graph is actually correct---everything else is just straight lines connecting the dots, which is our plotting software trying to **interpolate** (i.e., to approximate or guess the intermediate values) that lie between the points in the arrays we provided.

### Background subtraction

As we mentioned earlier, the experimental data seems to have a large background. That is, the spectrum only comes down to ~500000 instead of 0. There are many experimental issues that could give rise to such a effect (background light, improper grounding of the light detector, leaking electrons in the detector, etc.). What's important here is whether or not we are confident that this is a measurement issue instead of "real" signal. Let's assume here that we are confident and want to remove this part of the signal. 

We need to identify a region of the spectrum without any lines. Here, we will select the wavelength range from 560 nm to 600 nm. Our goal is to calculate the average signal in that region and subtract that average from the entire intensity. Our steps are:
1. Find the index of the wavelength array closest to 560 nm (index_1).
2. Find the index of the wavelength array closest to 600 nm (index_2).
3. Compute the average of the intensity array between index_1 and index_2.
4. Subtract that average.

Finding the index for the value closest to a wavelength requires us to define a new function:

```python
def find_index(array, value):
    return np.abs(array-value).argmin()
```

### For-Loops
Loops are a way of repeating the same operation over and over, which is useful for analyzing large datasets in an automated way. There are two primary looping methods in Python: "for-loops" and "while-loops". We'll focus on for-loops, which are more commonly used.

A for-loop allows you to specify what's known as an **iterator** — often an increasing array of indices — which lets you run a block of code over and over again under slightly different circumstances each time. There are many different ways to create an iterator.

You can use for loops to cycle through a list:

```python
# List of strings
list_1 = ["This", " is", " a", " list."]
for word in list_1:
    print(word)
```

```python
# List of values
list_2 = [2, 4, 6, 8, 10]
for num in list_2:
    print(num)
```

You can use the `range` function to create an iterator to access elements of a list or array:

```python
for i in range(len(list_1)):
    print(list_1[i])

for j in range(len(list_2)):
    print(list_2[j])
```

You have to be careful when using this method. If you try to access a list or array element that does not exist, Python will crash:

```python
for i in range(len(list_2)):
    print(list_1[i])
```

There are applications where we might need to write a for loop inside of a for loop. Those are called nested loops and we have to be very careful to not overdo it. Nesting for loops is a great way run out of memory. But they have their place and we'll be careful. For example, let's say you want to build a multiplication table up to 5x5. You could use nested loops:

```python
for num_1 in [1, 2, 3, 4, 5]:
    for num_2 in [1, 2, 3, 4, 5]:
        answer = num_1*num_2
        print(f"{num_1} x {num_2} = {answer}")
```

This can be a pretty compact way to do a lot of math.

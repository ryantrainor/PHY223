### Lists

Lists are the most flexible container in Python. A list holds multiple objects together, and each **element** of the list can be anything we want: ints, floats, strings, other data types. We can even create lists of lists! Of course, the usefulness of non-uniform containers becomes limited: the main advantage to storing lots of objects in a list is that you can then perform operations on them all without worrying that some won't work.

Here's how we define a list:

```python
list_1 = ['a string', 5, 6.0, True, [5,6,4]]
```

Notice the **square brackets** [ and ]: surrounding the elements above with square brackets *is one way to tell Python we are making a list*, rather than one of the other iterable types of Python objects that can contain different elements.

The list above is kind of a mess---you'd almost never want a list to contain such a variety of things in it, but we wanted to highlight that, in principle, Python doesn't care what you put in a list. In addition to manually specifiying what is in a list, we can use some **generative functions** to make lists for us when they follow a pattern or have a regular form:

```python
count_to_10 = list(range(1,11))
print(count_to_10)
```

What we've done above is run the `range()` function, which generates a series of numbers counting up. In Python 3, the `range()` function returns a special object called an **iterator**, but I can turn it into a list by using the `list()` function. *Recall from Part 1 that using a function to change the type of an object is called **typecasting**.*

Most of the functions we've seen so far have only a single argument, but the `range()` function can take up to three arguments inside its parentheses.

The form of the `range` function is `range(start, stop, step)`, where:
-  `start` represents the starting (integer) value
- `stop` represents the value that defines the end of the series, and
- `step` is the amount that each element should increase (allowing you to skip numbers).

Note that `start`  is inclusive and `stop` is exclusive---in mathematical notation, we have the set [1,11). This is why the list above starts at 1 and ends *before* 11.

If I wanted the numbers 0-9, I would do the following:

```python
list(range(10))
```

Note that we have used a shortcut here: if only 1 argument is used, `start` is assumed to be 0 and `step` is assumed to be 1. If two arguments are used, they are assumed to be `start` and `stop` (this is what we did when creating the list `count_to_10`).

The fact that the range output starts at zero and ends before the `stop` value can be confusing, but one good aspect to it is that `range(N)` will always return an iterable with `N` elements (starting at zero and ending at N-1). We can always check the length of a list (or other iterable) using the `len()` function:

```python
len(list_1)
```

### Indexing
When we want to see what is in a list, we print it. But sometimes we want to access individual elements of a list and use them in calculations. For that, we need to **index** or **slice** the list for the element(s) we want.

Lists are indexed such that the first element is assigned "0" (to remember this, some people call this element the "0th index"). Starting our indices at zero is part of Python that is different from some other languages (e.g., Java).

For example, the cell below will print only the first element in the list we defined above:

```python
list_1[0]
```

As you can see, indexing a list requires putting the (integer) index of the element we want within *square brackets* at the end of the variable name. The list we created has 5 elements, so we can index it with the integers 0, 1, 2, 3, and 4.

However, asking for index 5 will produce an error, since index 5 would be the 6th element in the array (which doesn't exist):

```python
list_1[5]
```

One useful trick is that we can also use **negative indexes** to *count backwards* from the end of the list, with -1 being the final element of the list, -2 being the second-to-last element, etc. This is useful when we don't know how long the list is, but we know how far the element we want is from the end:

```python
list_1[-2]
```

**Square brackets and parentheses are not interchangeable in Python!**

In particular, you should use **PARENTHESES** (and *not* square brackets) when you want to:

- "Call" a function (i.e., to use the function, placing any arguments to the function inside parentheses)
- Create a tuple (another type of iterable discussed below)

You should use **SQUARE BRACKETS** (and *not* parentheses) when you want to:
- Create a list
- Index an iterable to retrieve one or more elements stored inside it

## Slicing

Often times we want to access more than one element of a list (or another iterable object) without getting the whole object. Different types of iterables allow us to do this in different ways, but for lists our main option is to extract a **slice** of the list by specifying the starting and ending indices we want, separated by a colon (again inside *square brackets*):

```python
print(list_1)
print(list_1[0:2])
```

Notice that the syntax here is very similar to the `range(start,stop,skip)` command: the ending value is not included, which is why our slice included the 0 and 1 indexed elements, but not index 2. We can also specify a "skip" value:

```python
list_1[0:5:2]
```

Compare the output above to the full `list_1` to verify that the 0, 2, and 4 indexes (i.e., the first, third, and fifth elements) appeared.

So what about that list within our list? If we want a specific number from it, we can use *double indexing*.

```python
list_1[-1][1]
```

Again, compare the output to the full `list_1`: we first pulled out the element with the index -1 (i.e., the final element), which was the inner list, and then we indexed *that* for its own second element. _Remember: the second element is indexed with `[1]` because the first element is `[0]`!_

### Appending, extending, and duplicating lists

We have see that lists are flexible in terms of the types of information they can store and they way they can be indexed. Another useful property of lists is that it is easy to add additional data to them. We can add a single new element to the end of a list using the `.append()` method, or we can merge an entire list of new elements onto the end with the `.extend()` method:


```python
list_2 = [15, 'quartz', {'bounce':2}, 3]
list_2.append('Lancaster')
print(list_2)
```

```python
list_2 = [15, 'quartz', {'bounce':2}, 3]
list_2.extend(['Lancaster', 'York', 'Philadelphia'])
print(list_2)
```

Note that we can also use the `.append()` method to add a new list to the end of our original list, but it will have a slightly different result. Try it out below, and make sure you understand the difference in output between the two different versions!

```python
list_3 = list(range(5))
list_3.append([5, 6, 7])
print('Appended list:',list_3)

list_3 = list(range(5))
list_3.extend([5, 6, 7])
print('Extended list:',list_3)
```

Finally, note that the methods above actually change the list that has been stored in memory: after using the `.append()` or `.extend()` methods, the `list_3` object now has different contents. It can also be useful to use a list into a longer object in other ways---for example, lists can be added together or multiplied by integers in the same way as strings:

```python
abcs = ['a', 'b', 'c']
more_abcs = abcs + ['d', 'e', 'f']
repeated_abcs = more_abcs * 3
print(repeated_abcs)
```

Note that the cell above is different from appending and extending because each step actually redefines a new list instead of changing the original list `abcs`. However, all of these methods can be useful for manipulating lists.

### Dictionaries
Dictionaries store multiple values like lists, but rather than indexing them by element number as we were doing above, we index them by a special **key** to which we have assigned a **value**. For example:

```python
ages = {'Sam':5,'Sarah':6,'Kim':9,'Mukund':17}
ages['Sarah']
```

Dictionaries are inherently *unordered*---the order I define things within the dictionary doesn't matter, only the key associated with each value. Note that this means I can't index a dictionary with an integer index even if I want to! See what happens if I try to get the "first" age from the `ages` dictionary:

```python
ages[0]
```

The error message tells me my code had a "KeyError" because dictionaries expect to be indexed with a key, and 0 is not one of the keys in the `ages` dictionary: the only keys that are defined are 'Sam', 'Sarah', 'Kim' and 'Mukund'.

For some applications, dictionaries have advantages over lists. This is particularly true when we have values that are naturally paired with a unique identifier: if I have many students in a class who each have a final exam score, it makes sense that I would want to easily pull up a given student's score using their name, rather than having to remember whether they are the 5th or 8th or 31st student in my list of scores.

However, dictionaries cannot be easily sliced, it turns out that it is also more work to do math using dictionaries. For example, it would be harder to calculate the mean age from the `ages` dictionary above than it would be if we had the same ages in a list (or better yet, in a NumPy array, which we will discuss below).

One other note about dictionaries: similarly to lists, it is fairly easy to add new elements to them (i.e., new key : value pairs). We can do this by simply defining a new value to the dictionary with the key to which it should be assigned, like so:

```python
ages['Gregory'] = 72
print(ages)
```

### Tuples

We will not use tuples often, but it is good to know what they are. They have the special property of being **immutable**: once you define them, their values cannot be changed later in the code (unless you re-assign the variable name to something else or change the type). See the example below:

```python
example_list = [1,2,5]
example_tuple = (1,2,5)
example_list[0] = 2
print(example_list)
```

Notice that we have successfully defined a series of numbers as both a tuple and a list, and then we changed the 0th element of the list. But what if we try to change the 0th element of the tuple?

```python
example_tuple[0] = 2
```

We get an error. Similarly, tuples cannot be extended or have additial elements appended to them. If you make a list-type item and want to make sure it can never be adjusted in your code, you can make it a tuple. This is useful if you want to make sure a variable doesn't change unexpectedly!

### NumPy Arrays
Another type of iterable object we will use a lot is called an **array**, which is not part of basic Python. Rather, it is a new type of object that is we can use if we **import** an external **library** into Python to increase its functionality (one such library is NumPy, short for "Numerical Python"). We'll explore the limits of the native/basic Python tools and learn about NumPy arrays below.

Like all data types, NumPy arrays have pros and cons. They are less flexible than lists--all the elements of a NumPy array must be of the same type (float, int, string, etc.). We also cannot easily change the length of them by appending additional elements: a NumPy array is defined with a specific length, and it needs to be redefined in order to make it longer or shorter. However, we shall see that NumPy arrays are *much* more useful for data manipulation and mathematical operations than the other types we have discussed above!

## If Statements and other Conditionals
The problem of figuring out whether a condition is true or not is addressed in code via **conditional** statements. They run, logically, along the lines of "IF something is TRUE, do THIS, IF something ELSE is TRUE, do THAT, OTHERWISE do SOMETHING ELSE."

Here's an example:

```python
person_name = 'Finnaeus Fthrockbottom'
if len(person_name) < 10:
    print('This is a short name')
else:
    print('This is a long name')
```

So here, I asked whether the statement "the length of the string person_name is less than 10" was true or not. If it had been true, the first statement would've printed, but since it was not, the second one did.

You can also stack multiple if statements in a row:


```python
if 'F' in person_name:
    print('F Contained')
if 'n' in person_name:
    print('n Contained')
if 'l' in person_name:
    print('l Contained')
```

It prints that F and n are contained, but not for 'l', which is not. Here, all three if-statements are independent. But we can link them using elif statements, which are combinations of else and if:

```python
if 'F' in person_name:
    print('F Contained')
elif 'n' in person_name:
    print('n Contained')
```

What happened here? We said "If 'F' is in the name, print something, ELSE, IF it is NOT contained, but 'n' is contained, print something." But the first condition WAS met, so the else was never triggered. Thus, by mixing together if's, else's, and elif's, you can check conditions you are interested in. You can also ask multiple conditions in the same line:

```python
if ("F" in person_name) and ('n' in person_name):
    print('Both present')

if ("F" in person_name) or ('n' in person_name):
    print('One is present')

if ("l" not in person_name) and ('y' not in person_name):
    print('Neither present')

if ("l" not in person_name) or ('F' not in person_name):
    print('One is not present')
```

Take a moment parsing the above, seeing how you can string together conditionals. You can ask if things are in, or not in, other things, or you can compare values by asking if things are equal (==), not equal (!=), or greater than/less than (>, <).
